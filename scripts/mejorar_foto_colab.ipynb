```python
# 2 · Modelo: pesos oficiales de Real-ESRGAN, cargados con spandrel
!pip -q install spandrel
!wget -q -nc https://github.com/xinntao/Real-ESRGAN/releases/download/v0.1.0/RealESRGAN_x4plus.pth

import numpy as np, torch
from spandrel import ModelLoader

dev = torch.device("cuda" if torch.cuda.is_available() else "cpu")
if dev.type != "cuda":
    print("AVISO: sin GPU. Funciona, pero tarda unos 10-15 min. Entorno de ejecución → Cambiar tipo → GPU T4.")
model = ModelLoader().load_from_file("RealESRGAN_x4plus.pth").to(dev).eval()
print("Modelo cargado en", dev, "· escala x%d" % model.scale)
```

```python
# 3 · Reescalado x4 por teselas (cada tesela lleva un margen de contexto para que no se noten las juntas)
def upscale(model, img, tile=512, pad=24):
    s = model.scale
    h, w = img.shape[-2:]
    out = torch.empty(1, 3, h * s, w * s)
    for y in range(0, h, tile):
        for x in range(0, w, tile):
            y0, y1 = max(y - pad, 0), min(y + tile + pad, h)
            x0, x1 = max(x - pad, 0), min(x + tile + pad, w)
            with torch.no_grad(), torch.autocast(dev.type, enabled=dev.type == "cuda"):
                sr = model(img[..., y0:y1, x0:x1].to(dev)).float().cpu()
            th, tw = (min(y + tile, h) - y) * s, (min(x + tile, w) - x) * s
            ty, tx = (y - y0) * s, (x - x0) * s
            out[..., y * s:y * s + th, x * s:x * s + tw] = sr[..., ty:ty + th, tx:tx + tw]
    return out.clamp(0, 1)

x = torch.from_numpy(np.asarray(src)).permute(2, 0, 1).float().div(255).unsqueeze(0)
y = upscale(model, x)
x4 = Image.fromarray((y[0].permute(1, 2, 0).numpy() * 255).round().astype("uint8"))
print("Reescalada:", x4.size)
```

```python
# 4 · Se guarda al doble del original (de sobra para pantallas retina) y se compara la cara antes / después
x2 = x4.resize((src.width * 2, src.height * 2), Image.LANCZOS)
x2.save("lou-hero-x2.png", optimize=True)

box = (780, 20, 1100, 340)
before = src.crop(box).resize((640, 640), Image.LANCZOS)
after = x2.crop(tuple(v * 2 for v in box))
side = Image.new("RGB", (1290, 640), "white")
side.paste(before, (0, 0)); side.paste(after, (650, 0))
display(side)
```
